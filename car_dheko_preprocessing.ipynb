parse means converting a string (which looks like a dictionary or JSON) into an actual structured Python object, like a dictionary.
"{'fuel': 'Petrol', 'year': 2018}"  # This is a string
When you parse it using something like ast.literal_eval() or json.loads(), it becomes:
{'fuel': 'Petrol', 'year': 2018}  # Now it's a Python dictionary

Flatten means:
If that dictionary has nested data (like key-value pairs inside another dictionary), you “flatten” it by converting the nested structure into separate columns in a table.
{
  "user": {
    "id": 1,
    "info": {
      "name": "Rahul",
      "city": "Chennai"
    }
  }
}
from pandas import json_normalize

flat = json_normalize(data)
print(flat)
output:
{
  "user.id": 1,
  "user.info.name": "Rahul",
  "user.info.city": "Chennai"
}

```python
def parse_json_column(json_string):
    return ast.literal_eval(json_string) # ast.literal_eval() is a function in Python that evaluates a string containing a Python literal expression and returns the corresponding Python object.
```

```python
new_car_detail_b = df_banglore['new_car_detail'].apply(parse_json_column).apply(pd.Series)
new_car_overview_b = df_banglore['new_car_overview'].apply(parse_json_column).apply(pd.Series)

```

```python
new_car_detail_b
```

```python
new_car_detail_b.drop(["it","ft","transmission","owner","centralVariantId","variantName","priceActual","priceSaving","priceFixedText","trendingText"],axis=1,inplace=True)
```

```python
new_car_overview_b
```

```python
new_car_overview_b['top'] = new_car_overview_b['top'].apply(lambda lst: {d['key']: d['value'] for d in lst})
new_car_overview_b = new_car_overview_b['top'].tolist()
new_car_overview_b = pd.DataFrame(new_car_overview_b)
```

```python
new_car_overview_b
```

```python
new_car_overview_b.drop(["Registration Year","Insurance Validity","Kms Driven","RTO","Ownership","Engine Displacement","Transmission","Year of Manufacture"],axis=1,inplace= True)
```

```python
banglore_df= pd.concat([new_car_detail_b,new_car_overview_b],axis =1)
```

```python
banglore_df.info()
```

```python
banglore_df["city"] = "Banglore"
```

```python
banglore_df
```

```python
Column_names= ["Body_Type","Kilometers","Owners","Brand","Model","Model_Year","Price(Lakhs)","Fuel_Type","Seats","City"]
```

```python
df.columns= Column_names
```

```python
df.info()
```

```python
df.dropna(inplace= True)
```

```python
df["Kilometers"]= df["Kilometers"].str.replace(",", "").astype(int)
```

```python
df["Seats"]= df["Seats"].str.replace("Seats", "").astype(int) 
```

```python
df["Price(Lakhs)"] = df["Price(Lakhs)"].str.replace("₹", "", regex=False)
```

```python
def convert_price_to_lakhs(price_str):
    if "Crore" in price_str:
        return float(price_str.replace("Crore", "")) * 100
    elif "Lakh" in price_str:
        return float(price_str.replace("Lakh", ""))
    else:
        return None
```

```python
df["Price(Lakhs)"] = df["Price(Lakhs)"].apply(convert_price_to_lakhs)
```

```python
df.dropna(inplace=True)
```

```python
df.to_csv("final.csv",index=False)
```
